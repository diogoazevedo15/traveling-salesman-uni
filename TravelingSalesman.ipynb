```python
def traveling_salesman(T,alpha,graph,start):
    
    max_iteration = T/alpha
    tour = create_random_tour(graph,start)
    savedPaths = []
    
    while T - alpha >= 0:
        
        #Create new tour
        newTour = swap_random_elements(tour)
        
        #Calculate tour sums
        oldTourSum = get_tour_sum(tour)
        newTourSum = get_tour_sum(newTour)
        
        print('tour: ',tour)
        print('oldTourSum: ',oldTourSum)
        print('newTour: ',newTour)
        print('newTourSum: ',newTourSum)
        
        if random.random() <= T and newTourSum < oldTourSum:
            savedPaths.append(tour)
            tour = newTour
        T -= alpha
        
        print('--------------------------------------------------')
        
    print(savedPaths)
    print('Best path found: ',tour)
```

### Funções auciliares

```python
def create_random_tour(graph,start):
    
    edges = get_edges(graph,start) #['A','B','C'] start ='L'
    random.shuffle(edges) #['C','B','A']
    tour = [start] + edges + [start] #['L',C','B','A','L']

    return tour
```

```python
def get_edges(graph,start):
    
    edges = []
    for a,b in graph:
        if a not in edges and a != start:
            edges.append(a)
    return edges
```

```python
def get_tour_sum(tour):
    
    edges = [] 
    for i in range(len(tour)-1):
        edges.append((tour[i], tour[i+1]))
    
    soma = 0
    for i in edges:
        soma += graph[i]
    
    return soma
```
